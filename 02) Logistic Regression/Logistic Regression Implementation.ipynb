# **Explanation**
# Logistic Regression

Logistic regression is used for **binary classification**, where we apply the **sigmoid function**. The sigmoid function takes input as independent variables and produces a probability value between **0 and 1**.

For example, if we have two classes, **Class 0** and **Class 1**:
- If the value of the sigmoid function for an input is greater than **0.5** (threshold value), it belongs to **Class 1**.
- Otherwise, it belongs to **Class 0**.

It’s referred to as "regression" because it is an extension of linear regression, but it is mainly used for **classification problems**.

## Key Points:

- Logistic regression predicts the output of a **categorical dependent variable**. Therefore, the outcome must be a **categorical or discrete value**.
- The output can be values like **Yes or No**, **0 or 1**, **True or False**, etc. However, instead of giving exact values like 0 and 1, it provides **probabilistic values** that lie between **0 and 1**.
- In logistic regression, instead of fitting a regression line, we fit an **“S”-shaped logistic function**, which predicts two maximum values (0 or 1).

## Logistic Function – Sigmoid Function

- The **sigmoid function** is a mathematical function used to map the predicted values to **probabilities**.
- It maps any real value to another value within the range of **0 and 1**.
- The output of logistic regression must always be between **0 and 1**, which creates a curve in the shape of an **“S”**.
  - This S-shaped curve is called the **Sigmoid Function** or the **Logistic Function**.

- In logistic regression, we use the concept of a **threshold value**, which determines the probability of either **0 or 1**:
  - Values **above the threshold** tend to **1**.
  - Values **below the threshold** tend to **0**.

## Types of Logistic Regression

Logistic Regression can be classified into three types based on the categories of the dependent variable:

1. **Binomial**:
   - In binomial logistic regression, the dependent variable can have only **two possible types**, such as:
     - **0 or 1**
     - **Pass or Fail**
     - **Yes or No**

2. **Multinomial**:
   - In multinomial logistic regression, the dependent variable can have **3 or more possible unordered types**, such as:
     - **Cat, Dog, or Sheep**

3. **Ordinal**:
   - In ordinal logistic regression, the dependent variable can have **3 or more possible ordered types**, such as:
     - **Low, Medium, or High**

## Assumptions of Logistic Regression

Understanding the assumptions of logistic regression is crucial to ensure the appropriate application of the model. These assumptions include:

1. **Independent Observations**:
   - Each observation is **independent** of the others, meaning there is no correlation between any input variables.

2. **Binary Dependent Variable**:
   - The dependent variable must be **binary or dichotomous**, meaning it can take only two values.
   - For more than two categories, **SoftMax functions** are used.

3. **Linearity Between Independent Variables and Log Odds**:
   - The relationship between the independent variables and the **log odds** of the dependent variable should be **linear**.

4. **No Outliers**:
   - The dataset should not contain **outliers**, as they can affect the model's performance.

5. **Large Sample Size**:
   - The sample size should be **sufficiently large** to ensure reliable results.


```python
class LogisticRegression:

    def __init__ (self, learning_rate = 0.01, epochs = 1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.cost_history = []

    def sigmoid(self, z):
        z = np.clip(z, -500, 500)  # Prevent overflow
        return 1 / (1 + np.exp(-z))

    def cost_function(self, X, y):
        m = len(y)
        y_pred = self.sigmoid(np.dot(X, self.weights) + self.bias)

        # Clamp predicted values to avoid log(0)
        y_pred = np.clip(y_pred, 1e-10, 1 - 1e-10)

        cost = (-1 / m) * np.sum(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
        return cost

    def gradient_descent(self, X, y):
        
        m = len(y)
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for epoch in range(self.epochs):

            # Computer prediction
            y_pred = self.sigmoid(np.dot(X, self.weights) + self.bias)

            # Compute gradients for weights and bias
            dm = (1/m) * np.dot(X.T, (y_pred - y))
            db = (1/m) * np.sum(y_pred - y)

            # Update weights and bias
            self.weights -= self.learning_rate * dm
            self.bias -= self.learning_rate * db

            cost = self.cost_function(X, y)
            self.cost_history.append(cost)

            if epoch % 100 == 0:
                print(f'Epoch {epoch}, Cost {cost:.4f}')

    def fit(self, X, y):

        self.gradient_descent(X, y)
        return self.cost_history

    def predict(self, X):

        predictions = self.sigmoid(np.dot(X, self.weights) + self.bias)
        return (predictions >= 0.5).astype(int)
```

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
X = data.data
y = data.target
```

```python
model = LogisticRegression()
model.gradient_descent(X, y)
```
